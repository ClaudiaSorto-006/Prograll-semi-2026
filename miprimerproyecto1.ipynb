```python
# ==========================================
# SISTEMA ESTADISTICO ACADEMICO
# ==========================================

import json
import math


# ==========================================
# 1. DATOS EN JSON
# ==========================================

datos_json = '''
[
    {"nombre":"Ana", "materia":"Programacion", "nota":8.5, "edad":19, "traslado":30, "telefono":5},
    {"nombre":"Carlos", "materia":"Matematica", "nota":7.5, "edad":20, "traslado":40, "telefono":6},
    {"nombre":"Maria", "materia":"Programacion", "nota":9.0, "edad":19, "traslado":25, "telefono":4},
    {"nombre":"Jose", "materia":"Matematica", "nota":8.0, "edad":21, "traslado":50, "telefono":7},
    {"nombre":"Sofia", "materia":"Programacion", "nota":9.0, "edad":20, "traslado":35, "telefono":5},
    {"nombre":"Luis", "materia":"Matematica", "nota":7.0, "edad":22, "traslado":60, "telefono":8},
    {"nombre":"Daniela", "materia":"Programacion", "nota":8.5, "edad":19, "traslado":30, "telefono":6},
    {"nombre":"Pedro", "materia":"Matematica", "nota":7.5, "edad":23, "traslado":45, "telefono":7},
    {"nombre":"Laura", "materia":"Programacion", "nota":9.0, "edad":20, "traslado":20, "telefono":4},
    {"nombre":"Miguel", "materia":"Matematica", "nota":8.0, "edad":21, "traslado":55, "telefono":6},
    {"nombre":"Valeria", "materia":"Programacion", "nota":8.5, "edad":19, "traslado":30, "telefono":5},
    {"nombre":"Andres", "materia":"Matematica", "nota":7.0, "edad":22, "traslado":40, "telefono":8},
    {"nombre":"Gabriela", "materia":"Programacion", "nota":9.0, "edad":20, "traslado":35, "telefono":5},
    {"nombre":"Fernando", "materia":"Matematica", "nota":7.5, "edad":24, "traslado":65, "telefono":7},
    {"nombre":"Paola", "materia":"Programacion", "nota":8.5, "edad":21, "traslado":30, "telefono":4},
    {"nombre":"Ricardo", "materia":"Matematica", "nota":8.0, "edad":22, "traslado":50, "telefono":6},
    {"nombre":"Camila", "materia":"Programacion", "nota":9.0, "edad":19, "traslado":25, "telefono":5},
    {"nombre":"Diego", "materia":"Matematica", "nota":7.5, "edad":23, "traslado":60, "telefono":8},
    {"nombre":"Elena", "materia":"Programacion", "nota":8.0, "edad":20, "traslado":40, "telefono":6},
    {"nombre":"Roberto", "materia":"Matematica", "nota":8.5, "edad":21, "traslado":45, "telefono":7}
]
'''

# Convertimos el JSON a una lista de diccionarios
datos = json.loads(datos_json)


# ==========================================
# 2. CLASE PARA POO
# ==========================================

class Estadistica:

    def __init__(self, datos):
        self.datos = datos

    # --------------------------------------
    # Obtener los valores
    # --------------------------------------

    def obtener_valores(self, campo):

        valores = []

        for estudiante in self.datos:
            valores.append(estudiante[campo])

        return valores


# Creamos el objeto
estadistica = Estadistica(datos)


# ==========================================
# 3. FUNCION MEDIA
# ==========================================

def calcular_media(lista):

    suma = 0

    for numero in lista:
        suma = suma + numero

    media = suma / len(lista)

    return media


# ==========================================
# 4. FUNCION MEDIANA
# ==========================================

def calcular_mediana(lista):

    lista = sorted(lista)

    cantidad = len(lista)

    if cantidad % 2 == 0:

        mitad1 = lista[cantidad // 2 - 1]
        mitad2 = lista[cantidad // 2]

        mediana = (mitad1 + mitad2) / 2

    else:

        mediana = lista[cantidad // 2]

    return mediana


# ==========================================
# 5. FUNCION MODA
# ==========================================

def calcular_moda(lista):

    mayor = 0
    moda = None

    for numero in lista:

        cantidad = lista.count(numero)

        if cantidad > mayor:

            mayor = cantidad
            moda = numero

    if mayor == 1:

        return "No existe moda"

    return moda


# ==========================================
# 6. FUNCION VARIANZA
# ==========================================

def calcular_varianza(lista):

    media = calcular_media(lista)

    suma = 0

    for numero in lista:

        diferencia = numero - media

        suma = suma + diferencia ** 2

    varianza = suma / len(lista)

    return varianza


# ==========================================
# 7. FUNCION DESVIACION ESTANDAR
# ==========================================

def calcular_desviacion_estandar(lista):

    varianza = calcular_varianza(lista)

    desviacion = math.sqrt(varianza)

    return desviacion


# ==========================================
# 8. FUNCION RANGO
# ==========================================

def calcular_rango(lista):

    mayor = max(lista)
    menor = min(lista)

    rango = mayor - menor

    return rango


# ==========================================
# 9. FUNCION FRECUENCIA
# ==========================================

def calcular_frecuencia(lista):

    frecuencias = {}

    for numero in lista:

        if numero in frecuencias:

            frecuencias[numero] = frecuencias[numero] + 1

        else:

            frecuencias[numero] = 1

    return frecuencias


# ==========================================
# 10. MOSTRAR ESTADISTICAS
# ==========================================

def mostrar_estadisticas(campo):

    # Obtener los valores
    lista = estadistica.obtener_valores(campo)

    print("\n================================")
    print("RESULTADOS ESTADISTICOS")
    print("================================")

    print("Datos:", lista)

    print("Media aritmetica:",
          round(calcular_media(lista), 2))

    print("Mediana:",
          round(calcular_mediana(lista), 2))

    print("Moda:",
          calcular_moda(lista))

    print("Varianza:",
          round(calcular_varianza(lista), 2))

    desviacion = calcular_desviacion_estandar(lista)

    print("Desviacion estandar:",
          round(desviacion, 2))

    # Desviacion tipica es igual
    # a la desviacion estandar

    print("Desviacion tipica:",
          round(desviacion, 2))

    print("Rango:",
          round(calcular_rango(lista), 2))

    print("\nFRECUENCIA")

    frecuencias = calcular_frecuencia(lista)

    for valor in sorted(frecuencias):

        print(valor, "->", frecuencias[valor])


# ==========================================
# 11. MOSTRAR LOS ESTUDIANTES
# ==========================================

def mostrar_estudiantes():

    print("\n================================")
    print("LISTA DE ESTUDIANTES")
    print("================================")

    for estudiante in datos:

        print(
            estudiante["nombre"],
            "|", estudiante["materia"],
            "| Nota:", estudiante["nota"],
            "| Edad:", estudiante["edad"],
            "| Traslado:", estudiante["traslado"],
            "min | Telefono:", estudiante["telefono"],
            "horas"
        )


# ==========================================
# 12. MENU PRINCIPAL
# ==========================================

while True:

    print("\n================================")
    print(" SISTEMA ACADEMICO")
    print("================================")

    print("1. Ver estudiantes")
    print("2. Estadisticas de notas")
    print("3. Estadisticas de edades")
    print("4. Estadisticas de traslado")
    print("5. Estadisticas del telefono")
    print("6. Salir")

    opcion = input("Seleccione una opcion: ")

    if opcion == "1":

        mostrar_estudiantes()

    elif opcion == "2":

        mostrar_estadisticas("nota")

    elif opcion == "3":

        mostrar_estadisticas("edad")

    elif opcion == "4":

        mostrar_estadisticas("traslado")

    elif opcion == "5":

        mostrar_estadisticas("telefono")

    elif opcion == "6":

        print("Programa terminado.")
        break

    else:

        print("Opcion incorrecta.")
```

```python
class estudiantes:
    def __init__(self,nombre,notas,materias,edad,t_pantalla,t_universidad):
        self.nombre = nombre
        self.notas = notas
        self.materias = materias
        self.t_pantalla = t_pantalla
        self.t_universidad = t_universidad
        self.edad = edad

#MEDIA ARITMETICA
def calcular_media_aritmetica(estudiantes):
    sum_notas = sum(estudiantes.notas)
    total_notas = len(estudiantes.notas)
    resultado = sum_notas/total_notas


    return resultado

#key compara frecuencias y la funcion max por si sla no puede hacerlo
#MODA
def calcular_moda(estudiantes):
    notas = estudiantes.notas
    val_unico = set(notas)
    moda = max(val_unico , key=notas.count)

    return moda

#VARIANZA
def calcular_varianza(estudiantes):
    cantidad_notas = len(estudiantes.notas)
    promedio = sum(estudiantes.notas) / cantidad_notas

    suma_cuadrado = 0
    for elemento in estudiantes.notas:
        resta = elemento - promedio
        suma_cuadrado += resta ** 2

    varianza = suma_cuadrado / cantidad_notas

    return varianza

#RANGO
def calcular_rango(estudiantes):
    rango = max(estudiantes.notas)-min(estudiantes.notas)

    return rango

#para guardar los datos de los estudiantes
list_estudiante = [estudiantes("Jennifer Maricela Renderos Guevara",[10,10,10,10],
                ["Circuitos Digitales","Estadistica computacional",
                 "Administracion de bases de datos III","Ingles Basico"],21,
                 3,1),
                 estudiantes("Samuel Alexander Portillo Cerón",[10,10,10,10],
                    ["Circuitos Digitales","Estadistica computacional",
                     "Administracion de bases de datos III","Ingles Basico"],20,1.20,6),
                estudiantes("Carlos David Turcios Rivas",[9,9,10,10],
                     ["Circuitos Digitales","Estadistica computacional",
                        "Administracion de bases de datos III","Ingles Basico"],20,0.25,5),
                estudiantes("Gilda Mileidy Gónzales Castro",[10,10,10,10],
                     ["Circuitos Digitales","Estadistica computacional",
                        "Administracion de bases de datos III","Ingles Basico"],19,0.30,5),
                estudiantes("Edgar Ernesto ramirez Barrera",[10,10,10,10],
                     ["Circuitos Digitales","Estadistica computacional",
                         "Administracion de bases de datos III","Ingles Basico"],21,2,7),
                estudiantes("Yeimy Julissa Parada Chávez",[10,10,10,9],
                     ["Circuitos Digitales","Estadistica computacional",
                      "Administracion de bases de datos III","Ingles Basico"],20,0.45,3),
                estudiantes("Miguel Alejandro Goméz Saravia",[10,10,10,9],
                       ["Circuitos Digitales","Estadistica computacional",
                        "Administracion de bases de datos III","Ingles Basico"],19,0.15,4,),
                estudiantes("Josue Stiven Funes Benavides",[10,10,10,9],
                            ["Circuitos Digitales","Estadistica computacional",
                             "Administracion de bases de datos III","Ingles Basico"],19,0.30,5),
                estudiantes("Jefferson Stiven Castillo Leiva",[10,10,10,9],
                            ["Circuitos Digitales","Estadistica computacional",
                             "Administracion de bases de datos III","Ingles Basico"],19,0.15,2),
                estudiantes("Ramael Ernesto Cordova Colindres",[10,10,10,9],
                            ["Circuitos Digitales","Estadistica computacional",
                             "Administracion de bases de datos III","Ingles Basico"],22,0.7,13),
                estudiantes("Carlos Daniel Parada Avalos",[10,10,10,9],
                            ["Circuitos Digitales","Estadistica computacional",
                             "Administracion de bases de datos III","Ingles Basico"],25,0.10,5),
                estudiantes("Madisson Odalis Valle Juárez",[10,10,10,10],
                            ["Circuitos Digitales","Estadistica computacional",
                             "Administracion de bases de datos III","Ingles Basico"],19,0.30,9),
                estudiantes("Jose Luis Garcia Mejia",[10,10,10,10],
                            ["Circuitos Digitales","Estadistica computacional",
                             "Administracion de bases de datos III","Ingles Basico"],20,1.30,8)]


print("<<<<Sistema Estadístico Académico>>>>>")

#utilizare esto en caso de crear el menu de operaciones
operacion = ['Media aritmetica',
#Mediana
'Moda',
'Varianza',
#Desviacion estandar,
#Desviacion tipica ,
'Rango',
#Frecuencia


]


for i, persona in enumerate(list_estudiante, start=1):
    print(f"--- Persona #{i} ---")
    print(f"Nombre:    {persona.nombre}")
    print(f"Edad:      {persona.edad}")
    print("-" * 20)
estudiante=int(input("Selecciona un estudiante: "))-1
est_seleccionado = list_estudiante[estudiante]



diccionario_de_valores = {
   "moda":calcular_moda(est_seleccionado),
    "media aritmetica":calcular_media_aritmetica(est_seleccionado),
    "Varianza":calcular_varianza(est_seleccionado),
    "Rango":calcular_rango(est_seleccionado)}


formato = ""
for resultados,valores in diccionario_de_valores.items():
    l =  f"• {resultados}: {valores} \n"
    formato  +=  l
    print(f"• {resultados}: {valores}")
    # ==========================================
# SISTEMA ESTADISTICO ACADEMICO
# ==========================================

import json
import math


# ==========================================
# 1. DATOS EN JSON
# ==========================================

datos_json = '''
[
    {"nombre":"Ana", "materia":"Programacion", "nota":8.5, "edad":19, "traslado":30, "telefono":5},
    {"nombre":"Carlos", "materia":"Matematica", "nota":7.5, "edad":20, "traslado":40, "telefono":6},
    {"nombre":"Maria", "materia":"Programacion", "nota":9.0, "edad":19, "traslado":25, "telefono":4},
    {"nombre":"Jose", "materia":"Matematica", "nota":8.0, "edad":21, "traslado":50, "telefono":7},
    {"nombre":"Sofia", "materia":"Programacion", "nota":9.0, "edad":20, "traslado":35, "telefono":5},
    {"nombre":"Luis", "materia":"Matematica", "nota":7.0, "edad":22, "traslado":60, "telefono":8},
    {"nombre":"Daniela", "materia":"Programacion", "nota":8.5, "edad":19, "traslado":30, "telefono":6},
    {"nombre":"Pedro", "materia":"Matematica", "nota":7.5, "edad":23, "traslado":45, "telefono":7},
    {"nombre":"Laura", "materia":"Programacion", "nota":9.0, "edad":20, "traslado":20, "telefono":4},
    {"nombre":"Miguel", "materia":"Matematica", "nota":8.0, "edad":21, "traslado":55, "telefono":6},
    {"nombre":"Valeria", "materia":"Programacion", "nota":8.5, "edad":19, "traslado":30, "telefono":5},
    {"nombre":"Andres", "materia":"Matematica", "nota":7.0, "edad":22, "traslado":40, "telefono":8},
    {"nombre":"Gabriela", "materia":"Programacion", "nota":9.0, "edad":20, "traslado":35, "telefono":5},
    {"nombre":"Fernando", "materia":"Matematica", "nota":7.5, "edad":24, "traslado":65, "telefono":7},
    {"nombre":"Paola", "materia":"Programacion", "nota":8.5, "edad":21, "traslado":30, "telefono":4},
    {"nombre":"Ricardo", "materia":"Matematica", "nota":8.0, "edad":22, "traslado":50, "telefono":6},
    {"nombre":"Camila", "materia":"Programacion", "nota":9.0, "edad":19, "traslado":25, "telefono":5},
    {"nombre":"Diego", "materia":"Matematica", "nota":7.5, "edad":23, "traslado":60, "telefono":8},
    {"nombre":"Elena", "materia":"Programacion", "nota":8.0, "edad":20, "traslado":40, "telefono":6},
    {"nombre":"Roberto", "materia":"Matematica", "nota":8.5, "edad":21, "traslado":45, "telefono":7}
]
'''

# Convertimos el JSON a una lista de diccionarios
datos = json.loads(datos_json)


# ==========================================
# 2. CLASE PARA POO
# ==========================================

class Estadistica:

    def __init__(self, datos):
        self.datos = datos

    # --------------------------------------
    # Obtener los valores
    # --------------------------------------

    def obtener_valores(self, campo):

        valores = []

        for estudiante in self.datos:
            valores.append(estudiante[campo])

        return valores


# Creamos el objeto
estadistica = Estadistica(datos)


# ==========================================
# 3. FUNCION MEDIA
# ==========================================

def calcular_media(lista):

    suma = 0

    for numero in lista:
        suma = suma + numero

    media = suma / len(lista)

    return media


# ==========================================
# 4. FUNCION MEDIANA
# ==========================================

def calcular_mediana(lista):

    lista = sorted(lista)

    cantidad = len(lista)

    if cantidad % 2 == 0:

        mitad1 = lista[cantidad // 2 - 1]
        mitad2 = lista[cantidad // 2]

        mediana = (mitad1 + mitad2) / 2

    else:

        mediana = lista[cantidad // 2]

    return mediana


# ==========================================
# 5. FUNCION MODA
# ==========================================

def calcular_moda(lista):

    mayor = 0
    moda = None

    for numero in lista:

        cantidad = lista.count(numero)

        if cantidad > mayor:

            mayor = cantidad
            moda = numero

    if mayor == 1:

        return "No existe moda"

    return moda


# ==========================================
# 6. FUNCION VARIANZA
# ==========================================

def calcular_varianza(lista):

    media = calcular_media(lista)

    suma = 0

    for numero in lista:

        diferencia = numero - media

        suma = suma + diferencia ** 2

    varianza = suma / len(lista)

    return varianza


# ==========================================
# 7. FUNCION DESVIACION ESTANDAR
# ==========================================

def calcular_desviacion_estandar(lista):

    varianza = calcular_varianza(lista)

    desviacion = math.sqrt(varianza)

    return desviacion


# ==========================================
# 8. FUNCION RANGO
# ==========================================

def calcular_rango(lista):

    mayor = max(lista)
    menor = min(lista)

    rango = mayor - menor

    return rango


# ==========================================
# 9. FUNCION FRECUENCIA
# ==========================================

def calcular_frecuencia(lista):

    frecuencias = {}

    for numero in lista:

        if numero in frecuencias:

            frecuencias[numero] = frecuencias[numero] + 1

        else:

            frecuencias[numero] = 1

    return frecuencias


# ==========================================
# 10. MOSTRAR ESTADISTICAS
# ==========================================

def mostrar_estadisticas(campo):

    # Obtener los valores
    lista = estadistica.obtener_valores(campo)

    print("\n================================")
    print("RESULTADOS ESTADISTICOS")
    print("================================")

    print("Datos:", lista)

    print("Media aritmetica:",
          round(calcular_media(lista), 2))

    print("Mediana:",
          round(calcular_mediana(lista), 2))

    print("Moda:",
          calcular_moda(lista))

    print("Varianza:",
          round(calcular_varianza(lista), 2))

    desviacion = calcular_desviacion_estandar(lista)

    print("Desviacion estandar:",
          round(desviacion, 2))

    # Desviacion tipica es igual
    # a la desviacion estandar

    print("Desviacion tipica:",
          round(desviacion, 2))

    print("Rango:",
          round(calcular_rango(lista), 2))

    print("\nFRECUENCIA")

    frecuencias = calcular_frecuencia(lista)

    for valor in sorted(frecuencias):

        print(valor, "->", frecuencias[valor])


# ==========================================
# 11. MOSTRAR LOS ESTUDIANTES
# ==========================================

def mostrar_estudiantes():

    print("\n================================")
    print("LISTA DE ESTUDIANTES")
    print("================================")

    for estudiante in datos:

        print(
            estudiante["nombre"],
            "|", estudiante["materia"],
            "| Nota:", estudiante["nota"],
            "| Edad:", estudiante["edad"],
            "| Traslado:", estudiante["traslado"],
            "min | Telefono:", estudiante["telefono"],
            "horas"
        )


# ==========================================
# 12. MENU PRINCIPAL
# ==========================================

while True:

    print("\n================================")
    print(" SISTEMA ACADEMICO")
    print("================================")

    print("1. Ver estudiantes")
    print("2. Estadisticas de notas")
    print("3. Estadisticas de edades")
    print("4. Estadisticas de traslado")
    print("5. Estadisticas del telefono")
    print("6. Salir")

    opcion = input("Seleccione una opcion: ")

    if opcion == "1":

        mostrar_estudiantes()

    elif opcion == "2":

        mostrar_estadisticas("nota")

    elif opcion == "3":

        mostrar_estadisticas("edad")

    elif opcion == "4":

        mostrar_estadisticas("traslado")

    elif opcion == "5":

        mostrar_estadisticas("telefono")












    elif opcion == "6":

        print("Programa terminado.")
        break

    else:

        print("Opcion incorrecta.")
```
